### Step 38 — Create the Initial ML Feature Dataset

```python
ml_data = clean_data[
    [
        "year",
        "round",
        "raceId",
        "driverId",
        "constructorId",
        "circuitId",
        "grid",
        "qualifying_position",
        "podium"
    ]
].copy()

print("ML Dataset Shape:", ml_data.shape)

ml_data.head()
```

### Step 39 — Check ML Dataset Data Types

```python
print(ml_data.dtypes)
```

### Step 40 — Convert ML Features to Correct Numeric Types

```python
numeric_columns = [
    "year",
    "round",
    "raceId",
    "driverId",
    "constructorId",
    "circuitId",
    "grid",
    "qualifying_position",
    "podium"
]

for column in numeric_columns:
    ml_data[column] = pd.to_numeric(
        ml_data[column],
        errors="coerce"
    )

print(ml_data.dtypes)
```

### Step 41 — Check Missing Values After Numeric Conversion

```python
print(ml_data.isnull().sum())
```

### Step 42 — Check Missing Grid Records

```python
missing_grid = ml_data[
    ml_data["grid"].isnull()
]

missing_grid
```

### Step 43 — Remove Missing Grid Records

```python
ml_data = ml_data.dropna(
    subset=["grid"]
).copy()

print("ML Dataset Shape:", ml_data.shape)

print("\nMissing Values:")
print(ml_data.isnull().sum())
```

### Step 44 — Check Podium Target Distribution

```python
print(ml_data["podium"].value_counts())

print("\nPercentage:")
print(ml_data["podium"].value_counts(normalize=True) * 100)
```

### Step 45 — Define Features and Target

```python
X = ml_data[
    [
        "year",
        "round",
        "driverId",
        "constructorId",
        "circuitId",
        "grid",
        "qualifying_position"
    ]
].copy()

y = ml_data["podium"].copy()

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

### Step 46 — Split Training and Testing Data by Season

```python
train_mask = X["year"] <= 2024
test_mask = X["year"] >= 2025

X_train = X[train_mask].copy()
X_test = X[test_mask].copy()

y_train = y[train_mask].copy()
y_test = y[test_mask].copy()

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

## Step 47 — Import Machine Learning Libraries

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)
```

### Step 48 — Train Baseline Logistic Regression Model

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

### Step 49 — Make Podium Predictions

```python
y_pred = logistic_model.predict(X_test)

y_prob = logistic_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")
```

### Step 50 — Evaluate Model Accuracy

```python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Model Accuracy (%):", accuracy * 100)
```

### Step 51 — Generate Classification Report

```python
print(classification_report(y_test, y_pred))
```

### Step 52 — Generate Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

### Step 53 — Calculate ROC-AUC Score

```python
roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC Score:", roc_auc)
```

### Step 54 — Compare Actual vs Predicted Podium Results

```python
prediction_results = X_test.copy()

prediction_results["actual_podium"] = y_test.values
prediction_results["predicted_podium"] = y_pred
prediction_results["podium_probability"] = y_prob

prediction_results.head(20)
```

### Step 55 — Add Driver, Constructor and Race Names to Predictions

```python
prediction_results = prediction_results.merge(
    drivers[["driverId", "driver_name"]],
    on="driverId",
    how="left"
)

prediction_results = prediction_results.merge(
    constructors[["constructorId", "name"]],
    on="constructorId",
    how="left"
)

prediction_results.rename(
    columns={"name": "constructor_name"},
    inplace=True
)

prediction_results = prediction_results.merge(
    races[["raceId", "name"]],
    on="raceId",
    how="left"
)

prediction_results.rename(
    columns={"name": "race_name"},
    inplace=True
)

prediction_results.head()
```

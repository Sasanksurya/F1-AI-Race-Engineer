### Step 1 — Import Libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

### Step 2 — Load the Clean Dataset

```python
df = pd.read_csv("../Dataset/f1_clean_data.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

### Step 3 — Check Dataset Shape and Columns

```python
print("Dataset Shape:", df.shape)

print("\nTotal Columns:", len(df.columns))

print("\nColumns:")
print(df.columns.tolist())
```

#### Step 4 — Check Data Types

```python
df.dtypes
```

### Step 5 — Check Missing Values

```python
df.isnull().sum()
```

### Step 6 — Check Duplicate Records

```python
print("Duplicate Rows:", df.duplicated().sum())
```

### Step 7 — Check \N Missing Values

```python
missing_n = (df.astype(str) == r"\N").sum()

missing_n[missing_n > 0]
```

### Step 8 — Replace \N with NaN

```python
df = df.replace(r"\N", np.nan)

print("All \\N values replaced with NaN successfully!")
```

### Step 9 — Check Missing Values After Replacement

```python
df.isnull().sum()
```

### Step 10 — Convert Important Numeric Columns

```python
numeric_columns = [
    "grid",
    "position",
    "time",
    "milliseconds",
    "fastestLap",
    "rank",
    "fastestLapSpeed",
    "qualifying_position"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print("Numeric columns converted successfully!")
```

### Step 11 — Check Data Types After Conversion

```python
df[
    [
        "grid",
        "position",
        "time",
        "milliseconds",
        "fastestLap",
        "rank",
        "fastestLapSpeed",
        "qualifying_position"
    ]
].dtypes
```

### Step 12 — Remove Rows with Missing Grid Position

```python
df = df.dropna(
    subset=["grid"]
).copy()

print("Dataset Shape:", df.shape)
print("Missing Grid Values:", df["grid"].isnull().sum())
```

### Step 13 — Check Missing Values Percentage

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

missing_percentage[missing_percentage > 0]
```

### Step 14 — Remove Unnecessary High-Missing Columns

```python
columns_to_drop = [
    "time",
    "milliseconds"
]

df = df.drop(
    columns=columns_to_drop
)

print("Columns removed successfully!")
print("Dataset Shape:", df.shape)
```

### Step 15 — Check Remaining Missing Values

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

### Step 16 — Handle Remaining Missing Values

```python
df["position"] = df["position"].fillna(df["positionOrder"])
df["fastestLap"] = df["fastestLap"].fillna(df["fastestLap"].median())
df["rank"] = df["rank"].fillna(df["rank"].median())
df["fastestLapSpeed"] = df["fastestLapSpeed"].fillna(
    df["fastestLapSpeed"].median()
)

print("Remaining missing values handled successfully!")
```

### Step 17 — Verify Missing Values After Cleaning

```python
df.isnull().sum()[df.isnull().sum() > 0]
```

### Step 18 — Check Duplicates After Cleaning

```python
print("Duplicate Rows:", df.duplicated().sum())
```

### Step 19 — Save Final Cleaned Dataset

```python
df.to_csv(
    "../Dataset/f1_final_cleaned_data.csv",
    index=False
)

print("Final cleaned dataset saved successfully!")
print("Final Dataset Shape:", df.shape)
```

### Step 20 — Verify Saved Final Cleaned Dataset

```python
final_df = pd.read_csv(
    "../Dataset/f1_final_cleaned_data.csv"
)

print("Dataset Shape:", final_df.shape)
print("Duplicate Rows:", final_df.duplicated().sum())
print("Total Missing Values:", final_df.isnull().sum().sum())

final_df.head()
```
